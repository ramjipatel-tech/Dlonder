# Dlonder — Google Colab launcher

Run the Telegram bot from Google's Colab runtime. Your phone does not need to stay connected, but **Colab Free is not 24/7 hosting** and may disconnect or stop the runtime.

## Before running
1. Run the setup cell to install dependencies and download the repository.
2. In the next cell, paste your Telegram bot token when prompted. The input is hidden while typing and is not saved in this notebook.
3. Optionally enter your own authorized TeraBox session cookie when prompted. Leave it blank if you do not use TeraBox.

Never share your bot token or session cookie in chat, screenshots, or GitHub. Only download files you own or are authorized to access.

```python
# Enter credentials privately (input is hidden; values are not saved in the notebook)
import os
from getpass import getpass

bot_token = getpass('Enter Telegram BOT_TOKEN (input hidden): ').strip()
if not bot_token:
    raise ValueError('BOT_TOKEN is required. Run this cell again and enter your bot token.')
os.environ['BOT_TOKEN'] = bot_token

terabox_cookie = getpass('Optional: enter your own TERABOX_COOKIE (press Enter to skip): ').strip()
if terabox_cookie:
    os.environ['TERABOX_COOKIE'] = terabox_cookie

os.environ['MAX_FILE_MB'] = '45'
os.environ['DOWNLOAD_DIR'] = '/content/dlonder-downloads'
os.environ.setdefault('DISKWALA_RESOLVER_URL', 'https://diskwala-dl-six.vercel.app/api/scrap')

print('Credentials loaded. Starting Dlonder…')
!python -m bot.main
```
