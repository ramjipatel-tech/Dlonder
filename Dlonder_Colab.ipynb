# Dlonder — Google Colab launcher

Run the Telegram bot from Google's Colab runtime. Your phone does not need to stay connected, but **Colab Free is not 24/7 hosting** and may disconnect or stop the runtime.

## Before running
1. Open the **Secrets** panel (key icon on the left).
2. Add a secret named `BOT_TOKEN` with your Telegram bot token.
3. Enable notebook access to that secret.
4. Optionally add `TERABOX_COOKIE` as a secret if you use your own authorized TeraBox session. Never share it publicly.

Only download files you own or are authorized to access.

```python
# Get the latest public repository code
%cd /content
!rm -rf Dlonder
!git clone --depth 1 https://github.com/ramjipatel-tech/Dlonder.git
%cd /content/Dlonder
!pip -q install -r requirements.txt
```

```python
# Load credentials from Colab Secrets (never paste tokens into notebook cells)
import os
from google.colab import userdata

os.environ['BOT_TOKEN'] = userdata.get('BOT_TOKEN')
os.environ['MAX_FILE_MB'] = '45'
os.environ['DOWNLOAD_DIR'] = '/content/dlonder-downloads'
os.environ.setdefault('DISKWALA_RESOLVER_URL', 'https://diskwala-dl-six.vercel.app/api/scrap')

try:
    cookie = userdata.get('TERABOX_COOKIE')
    if cookie:
        os.environ['TERABOX_COOKIE'] = cookie
except Exception:
    pass

print('Credentials loaded. Starting Dlonder…')
!python -m bot.main
```
